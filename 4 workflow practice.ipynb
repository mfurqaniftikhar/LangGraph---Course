# Sequential Workflow

```python
from langgraph.graph import StateGraph, END, START

# create function and state graph
def sequentile_workflow():
    graph = StateGraph(dict)

    # add nodes
    graph.add_node("step1", lambda state: print("Step 1 done"))
    graph.add_node("step2", lambda state: print("Step 2 done"))
    graph.add_node("step3", lambda state: print("Step 3 done"))
    # graph.add_node("step 4", lambda)

    # edges
    graph.add_edge(START, "step1")
    graph.add_edge("step1", "step2")
    graph.add_edge("step2", "step3")
    graph.add_edge("step3", END)

    # compile and show graph
    add = graph.compile()
    add.invoke({})
sequentile_workflow()
```

# Parallel Workflow (multiple nodes ek saath)

```python
from langgraph.graph import StateGraph, START,END
# function create and stategraph
def parellel_workflow():
    graph = StateGraph(dict)


    # add nodes
    graph.add_node("taskA", lambda s: print("Task A Done"))
    graph.add_node("taskB", lambda s: print("Task B Done"))
    graph.add_node("taskC", lambda s : print("Task C Done"))
    
    # add adges
    # Start se teen parallel edges
    graph.add_edge(START, "taskA")
    graph.add_edge(START, "taskB")
    graph.add_edge(START, "taskC")
    
    # Har task END tak jata hai
    graph.add_edge("taskA", END)
    graph.add_edge("taskB", END)
    graph.add_edge("taskC", END)

    app = graph.compile()
    app.invoke({})
parellel_workflow()

    
```

# Componential Workflow (modular sub-graphs)

```python
from langgraph.graph import StateGraph, START, END
def componential_workflow():
    # Component 1: Data Processing
    
    data_graph = StateGraph(dict)
    data_graph.add_node("load", lambda s : print("Data Loaded"))
    data_graph.add_node("clean", lambda s : print("Clean Loaded"))
    data_graph.add_edge(START, "load")
    data_graph.add_edge("load", "clean")
    data_graph.add_edge("clean", END)
    data_component = data_graph.compile()
    
    # Component 2: Model Operations
    data_graph = StateGraph(dict)
    data_graph.add_node("Train", lambda s : print("Model Trained"))
    data_graph.add_node("eval", lambda s : print("Model Evaluated"))
    data_graph.add_edge(START, "Train")
    data_graph.add_edge("Train", "eval")
    data_graph.add_edge("eval", END)
    model_component=data_graph.compile()

    # Main workflow
    main_graph = StateGraph(dict)
    main_graph.add_node("data_component", lambda s : data_component.invoke({}))
    main_graph.add_node("model_component", lambda s : model_component.invoke({}))
    main_graph.add_edge(START, "data_component")
    main_graph.add_edge("data_component","model_component")
    main_graph.add_edge("model_component", END)

    # compile and run
    run = main_graph.compile()
    run.invoke({})
componential_workflow()

    
    
    
    
    
    
```

# Iterative Workflow (loop-based)

```python
from langgraph.graph import StateGraph, END, START
def iterative_workflow():
    graph = StateGraph(dict)
    
    # defines Nodes
    graph.add_node("initialize", lambda s: print("Iteration start"))
    graph.add_node("process", lambda s: print("procss  iteration{s['i']}"))
    graph.add_node("check", lambda s : print("Check condition"))

     # Edges for iteration
    graph.add_edge(START, "initialize")
    graph.add_edge("initialize", "process")
    graph.add_edge("process", "check")
    
    # Loop condition
    def check_condition(state):
        if state["i"] <3:
            state["i"] +=1
            return "Process"
        else:
            return END

    graph.add_conditional_edges("check", check_condition)

    app = graph.compile()
    app.invoke({"i":1})
iterative_workflow()

    
    
```
