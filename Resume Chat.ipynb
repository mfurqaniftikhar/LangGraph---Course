```python
from langgraph.graph import StateGraph, HumanMessage, AIMessage

# Step 1: Graph aur Memory create karo
graph = StateGraph()
memory = Memory()

# Step 2: User se input lo
user_input = "Hello AI, mera naam Muhammad hai."

# Step 3: Memory mai user ka message save karo
memory.add(HumanMessage(user_input))

# Step 4: AI ka node banao
ai_node = AIMessage("Hello Muhammad! Kaise ho?")

# Step 5: Memory mai AI ka response add karo
memory.add(ai_node)

# Step 6: Graph mai dono connect karo
graph.add_edge(memory.messages[0], memory.messages[1])

# Step 7: Graph run karo
graph.run()

# Step 8: Jab resume karna ho (e.g. next day)
new_input = "Main theek hoon, tum batao?"
memory.add(HumanMessage(new_input))

# AI continue karega purani memory ke sath
ai_next = AIMessage("Main bhi theek hoon! Hum kal yahan ruk gaye the.")
memory.add(ai_next)

# Print conversation
for msg in memory.messages:
    print(f"{msg.type}: {msg.content}")
```

```python
# save in json
# Non-persistent (RAM mai)
memory = {"chat": ["hello", "hi!"]},
# Persistent (file mai save)
import json

# Save
with open("memory.json", "w") as f:
    json.dump(memory, f)

# Load again (resume chat)
with open("memory.json", "r") as f:
    memory = json.load(f)

print(memory)
```

```python
from langgraph import Memory

# Suppose har chat ka apna thread_id hai
thread_id = "user123_thread5"

# Memory store (simulate DB)
all_threads = {}

# Resume check
if thread_id in all_threads:
    memory = all_threads[thread_id]
else:
    memory = Memory()
    all_threads[thread_id] = memory

# Chat flow
memory.add({"role": "user", "content": "Hello!"})
memory.add({"role": "ai", "content": "Hi there!"})

# Later resume
print("Resumed memory for:", thread_id)
for msg in memory.messages:
    print(msg)
```
