```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict,Literal, Annotated
from pydantic import BaseModel, Field
from langchain_groq import ChatGroq
```

```python
import langchain
import operator
from langchain_core.messages import HumanMessage, BaseMessage
from langgraph.graph.message import add_messages
from langgraph.checkpoint.memory import MemorySaver 
```

```python
model=ChatGroq(model='llama-3.1-8b-instant',api_key='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')
```

```python
class chatbot(TypedDict):
    message : Annotated[list[BaseMessage], add_messages]
    
```

```python
check_point = MemorySaver()
```

```python
graph = StateGraph(chatbot)
```

```python
def chat_model(state: chatbot):
    response = state['message']
    res  = model.invoke(response)
    return {"message": [res]}
```

```python
# add nodes
graph.add_node("chat_model", chat_model)
# add_edges
graph.add_edge(START, "chat_model")
graph.add_edge("chat_model", END)

# compile 
run = graph.compile(checkpointer=check_point)
```

```python
thread_id = "1"
while True:
    user_type = input("User Type\n")
    if user_type.lower() in ["exit", "bye"]:
        break
    config = {"configurable": {'thread_id' :thread_id}}
    ai = run.invoke({"message": HumanMessage(content=user_type)}, config=config)
    print("Ai message", ai["message"][-1].content)
```
